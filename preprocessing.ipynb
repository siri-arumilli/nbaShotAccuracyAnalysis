Check the Shape of the Dataset and if there are any missing values.

```python
# Our data has no missing values, verify by the output.
# There are no null values in any of the columns.
print(f"Dataset shape: {df.shape}")
print(f"Missing values:\n{df.isnull().sum()}")
```

We can get rid of the event_type data because it is redundant, shot_made is consistent with event_type.

```python
print("Checking to make sure EVENT_TYPE and SHOT_MADE have consistent data:")
inconsistent_made = df[(df['EVENT_TYPE'].str.contains('Made')) & (df['SHOT_MADE'] == False)]
inconsistent_missed = df[(df['EVENT_TYPE'].str.contains('Missed')) & (df['SHOT_MADE'] == True)]
print(f"Inconsistent 'Made Shot' rows: {len(inconsistent_made)}")
print(f"Inconsistent 'Missed Shot' rows: {len(inconsistent_missed)}")
```

```python
df_clean = df.drop('EVENT_TYPE', axis=1)
print(f"Removed column: EVENT_TYPE, it was redundant")
```
